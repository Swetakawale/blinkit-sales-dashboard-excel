HeatMap- Correlation of one variable with others for numerucal column

```python
corr = data[['latitude', 'longitude', 'price', 'minimum_nights',
             'number_of_reviews', 'reviews_per_month',
             'availability_365', 'beds']].corr()
corr
plt.figure(figsize=(8,8))
sns.heatmap(data=corr,annot=True)
```
